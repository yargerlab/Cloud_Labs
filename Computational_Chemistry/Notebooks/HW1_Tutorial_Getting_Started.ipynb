## Step 2: Working with SMILES in RDKit
**SMILES** (Simplified Molecular Input Line Entry System) represents chemical structures as short text strings.

Let's import `rdkit.Chem` and parse a SMILES string for **Aspirin** (`CC(=O)Oc1ccccc1C(=O)O`).

```python
from rdkit import Chem
from rdkit.Chem import Draw
from rdkit.Chem import rdMolDescriptors

# Define SMILES for Aspirin
smiles_aspirin = 'CC(=O)Oc1ccccc1C(=O)O'
mol_aspirin = Chem.MolFromSmiles(smiles_aspirin)

print(f'Molecule Formula: {rdMolDescriptors.CalcMolFormula(mol_aspirin)}')
print(f'Heavy Atom Count: {mol_aspirin.GetNumHeavyAtoms()}')

# Render 2D Structure
Draw.MolToImage(mol_aspirin, size=(300, 300))
```

## Step 3: Implicit vs. Explicit Hydrogens
By default, SMILES and 2D chemical graphs suppress hydrogen atoms to keep text concise. In quantum chemistry and 3D modeling, explicit hydrogens are required.

Let's compare Propane (`CCC`) with implicit vs. explicit hydrogens:

```python
mol_propane = Chem.MolFromSmiles('CCC')
mol_propane_explicit = Chem.AddHs(mol_propane)

print(f'Propane without AddHs atom count: {mol_propane.GetNumAtoms()}')
print(f'Propane with AddHs atom count: {mol_propane_explicit.GetNumAtoms()}')

# Render side-by-side
img = Draw.MolsToGridImage([mol_propane, mol_propane_explicit], legends=['Implicit H', 'Explicit H'], subImgSize=(250,250))
img
```

## Step 4: Generating 3D Conformers & Interactive Rendering
SMILES strings do not store 3D spatial coordinates. We generate 3D structures using distance geometry algorithms (**ETKDGv3**) and optimize geometries with empirical force fields (**MMFF94**).

```python
from rdkit.Chem import AllChem
import py3Dmol

# 1. Add Hydrogens
mol_3d = Chem.AddHs(mol_aspirin)

# 2. Embed 3D Conformer (ETKDG)
AllChem.EmbedMolecule(mol_3d, AllChem.ETKDGv3())

# 3. Optimize Geometry with MMFF94 force field
AllChem.MMFFOptimizeMolecule(mol_3d)

# 4. Render interactively in 3D using py3Dmol
mb = Chem.MolToMolBlock(mol_3d)
view = py3Dmol.view(width=400, height=400)
view.addModel(mb, 'mol')
view.setStyle({'stick': {}, 'sphere': {'scale': 0.25}})
view.zoomTo()
view.show()
```

## Step 5: Biological Macromolecules (FASTA & Ribbon Views)
For large biological peptides and proteins, **FASTA strings** are used instead of SMILES, and 3D visualization uses **ribbon diagrams** rather than explicit atom spheres.

```python
# Load Human Insulin 3D Structure from RCSB PDB (PDB ID: 1TRZ)
view_pdb = py3Dmol.view(query='pdb:1TRZ', width=500, height=400)
view_pdb.setStyle({'ribbon': {'color': 'spectrum'}})
view_pdb.zoomTo()
view_pdb.show()
```
